```python
import pandas as pd
import numpy as np

data = {
    "country": ["Brazil", "China", "India", "Russia", "South Africa", "USA", "Japan", "Germany"],
    "capital": ["Brasilia", "Beijing", "New Dehli", "Moscow", "Pretoria","Washington, D.C.", "Tokyo", "Berlin"],
    "population": [212, 1410, 1430, 144, 60, 340, 124, 84],
    "gdp_per_capita": [10000, 12500, 2700, 15000, 7000, 85000, 34000, 55000],
    "area": [8.5, 9.6, 3.3, 17.1, 1.2, 9.8, 0.38, 0.36]
}

countries = pd.DataFrame(data)

print(countries)
```

```python
index = 0

while countries.loc[index, "population"] <= 1000: 
    print(countries.loc[index, "country"])
    index += 1
```

```python
for index, country in enumerate(countries["country"]):
    print(index, country)
```

```python
population_dict = {
    "Brazil": 212,
    "China": 1410,
    "India": 1430,
    "Russia": 144,
    "South Africa": 60,
    "USA": 340,
    "Japan": 124,
    "Germany": 84
}

for country, population in population_dict.items():
    print(country, population)
```

```python
population_array = np.array(countries["population"])

for population in population_array:
    print(population)
```

```python
measurements = np.array([countries["population"], countries["area"]])

for row in measurements:
    print(row)
```

```python
for value in np.nditer(measurements):
    print(value)
```

```python
for index, row in countries.iterrows():
    print(index, row["country"], row["gdp_per_capita"])
```

```python
countries["name_length"] = 0

for index, row in countries.iterrows(): countries.loc[index, "name_length"] = len(row["country"])
print(countries)
```

```python
countries["name_length_apply"] = countries["country"].apply(len)
print(countries[["country", "name_length", "name_length_apply"]])
```
